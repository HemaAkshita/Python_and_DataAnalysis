```python
import copy
```

```python
numberList = [23, 34, 12, 25, 46, 35, 58, 28, 19, 65, 20, 48, 17, 68, 85, 93, 74, 33, 92, 61] #  [1, 1, 5, 5, 12, 12, 45, 45, 99, 99] 
# List of 20 numbers
```

```python
sum = 0
count = 0
max = -999999
min = 999999

for i in numberList:
    sum += i
    count += 1
    if (i > max):
        max = i
    if (min > i):
        min = i
        
avg = sum/count

print("Maximum = ", max, "\nMinimum = ", min, "\nSum = ", sum, "\nAverage = ", avg)
```

```python
def printEvenNumbers(numberList):
    for num in numberList:
        if not(num % 2):
            print(num)


def printOddNumbers(numberList):
    for num in numberList:
        if (num % 2):
            print(num)
            
```

```python
printEvenNumbers(numberList)
```

```python
printOddNumbers(numberList)
```
